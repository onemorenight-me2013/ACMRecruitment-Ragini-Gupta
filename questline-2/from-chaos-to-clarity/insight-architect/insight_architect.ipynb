```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


df=pd.read_csv("StudentsPerformance.csv")
df.head()
```

```python
# histogram for maths score
plt.figure(figsize=(8, 4))
sns.histplot(df['math score'], kde=True, color='skyblue')
plt.title("Distribution of Math Scores", fontsize=14, fontweight='bold')
plt.xlabel("Math Score")
plt.ylabel("Frequency")
plt.show()
```

```python
# Boxplot: Math Score vs Test Preparation
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='test preparation course', y='math score', palette='Set2')
plt.title("Math Score Distribution by Preparation Course", fontsize=14)
plt.show()
```

```python
plt.scatter(df["math score"], df["reading score"], alpha=0.5)
plt.xlabel("Math Score")
plt.ylabel("Reading Score")
plt.title("Math vs Reading Score")
plt.show()
```

```python
plt.figure(figsize=(7, 5))
corr = df[['math score', 'reading score', 'writing score']].corr()
sns.heatmap(corr, annot=True, cmap='Blues', fmt='.2f', linewidths=1)
plt.title("Correlation Matrix of Test Scores", fontsize=14)
plt.show()
```

```python
counts = df["gender"].value_counts()
plt.bar(counts.index, counts.values)
plt.title("Count by Gender")
plt.ylabel("Number of Students")
plt.show()
```
